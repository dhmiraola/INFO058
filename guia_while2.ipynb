## 1. Repaso: las tres partes de `while`
1. **Inicio:** ¿desde qué valor comenzamos?
2. **Condición:** ¿mientras qué sea verdadero repetimos?
3. **Cambio:** ¿qué hacemos para que eventualmente deje de cumplirse?

`numero += 1` significa lo mismo que `numero = numero + 1`.

````python
numero = 1  # Comenzamos en 1

while numero <= 5:  # Repetimos mientras numero sea 5 o menos
    print(numero)  # Mostramos el numero actual
    numero += 1  # Aumentamos el numero para avanzar

print("Terminó el ciclo")  # Se ejecuta al salir del while
````

## 2. La condición puede ser falsa desde el principio
Si la condición es falsa al entrar, el `while` no se ejecuta ninguna vez.

````python
numero = 10

while numero < 5:
    print(numero)
    numero += 1

print("El programa continúa")
````

## 3. ¿Cuándo aparece un bucle infinito?
Un **bucle infinito** ocurre cuando la condición continúa siendo verdadera y nada hace que cambie.

**Ejemplo para analizar, NO ejecutar:**
```python
numero = 1
while numero <= 5:
    print(numero)
```

Como `numero` nunca cambia, siempre vale 1. El programa seguirá imprimiendo 1.

**Corrección:** agregar `numero += 1` dentro del ciclo.

Si ejecutas accidentalmente un bucle infinito, usa **Interrumpir ejecución / Interrupt kernel** (o Ctrl+C en la terminal).

### Otro error común: avanzar en la dirección equivocada
**No ejecutar:**
```python
numero = 5
while numero >= 1:
    print(numero)
    numero += 1
```
El número crece, por lo que nunca será menor que 1. Para una cuenta regresiva, debemos **restar**.

````python
numero = 5

while numero >= 1:
    print(numero)
    numero -= 1  # Restamos 1 para acercarnos al final

print("Fin de la cuenta regresiva")
````

## 4. `while` + `if` y `else`
El `while` **repite**. El `if` **decide** qué mostrar en cada vuelta.

Aquí revisamos si un número es par o impar.

````python
numero = 1

while numero <= 5:
    if numero % 2 == 0:  # El resto de dividir por 2 es cero
        print(numero, "es par")
    else:
        print(numero, "es impar")

    numero += 1  # Avanzamos en cada vuelta
````

## 5. `while` + `if`, `elif` y `else`
Podemos elegir entre **tres resultados** distintos.

````python
numero = 1

while numero <= 6:
    if numero < 3:
        print(numero, "es pequeño")
    elif numero == 3:
        print(numero, "es igual a 3")
    else:
        print(numero, "es mayor que 3")

    numero += 1
````

## 6. `break`: salir inmediatamente del ciclo
`break` corta el `while` **en ese momento**, aunque la condición todavía sea verdadera.

````python
numero = 1

while numero <= 10:
    if numero == 4:
        print("Encontramos el 4. Detenemos el ciclo")
        break  # Salimos inmediatamente del while

    print(numero)
    numero += 1

print("Estamos fuera del while")
````

### Ejemplo con `while True`
`while True` no termina por su condición: necesitamos una salida, como `break`.

````python
while True:
    numero = int(input("Escribe un número (0 para salir): "))

    if numero == 0:
        print("Terminamos")
        break

    print("Ingresaste:", numero)
````

## 7. `continue`: saltar una vuelta, sin terminar el ciclo
`continue` **no termina** el `while`. Salta el resto de la vuelta actual y vuelve a comprobar la condición.

**Importante:** en un `while` con contador, debemos actualizarlo **antes** de `continue` si, de lo contrario, quedaría detenido en el mismo valor.

````python
numero = 0

while numero < 5:
    numero += 1  # Primero avanzamos: 1, 2, 3, 4, 5

    if numero == 3:
        continue  # Saltamos el print cuando numero vale 3

    print(numero)

print("Terminó el ciclo")
````

**Salida esperada:** 1, 2, 4, 5.

`continue` omite la impresión del 3, pero el ciclo continúa normalmente.

## 8. Comparación: `break` y `continue`
| Instrucción | ¿Qué hace? | ¿El ciclo sigue? |
|---|---|---|
| `break` | Sale del ciclo completo | No |
| `continue` | Salta lo que queda de la vuelta actual | Sí, si la condición sigue siendo verdadera |

**Regla fácil:** `break` = salir; `continue` = saltar esta vuelta.

## 9. Ejercicio guiado: clasificar números hasta que aparezca 0
**Enunciado:** pide números enteros. Si el usuario escribe 0, termina. Para los demás, indica si son positivos o negativos.

````python
while True:
    numero = int(input("Ingresa un número (0 para salir): "))

    if numero == 0:
        print("Fin del programa")
        break
    elif numero > 0:
        print("Positivo")
    else:
        print("Negativo")
````

## 10. Ejercicio guiado: ignorar el cero
**Enunciado:** pide 5 números. Muestra todos, excepto los ceros. Aquí usamos `continue`.

````python
contador = 0

while contador < 5:
    numero = int(input("Ingresa un número: "))
    contador += 1  # Contamos el ingreso, incluso si es cero

    if numero == 0:
        continue  # No mostramos el cero

    print("Número:", numero)

print("Ingresaste 5 números")
````

## 12. Resumen de la clase

- `while` repite instrucciones mientras una condición sea verdadera.
- El contador debe avanzar hacia el término del ciclo para evitar repeticiones infinitas.
- `if`, `elif` y `else` permiten tomar decisiones dentro del ciclo.

- `continue` salta el resto de la vuelta actual y pasa a la siguiente.
