### Ejercicio 3: Repetir un saludo

**Enunciado:** Imprime la palabra "Hola" 5 veces.

**Solución:**

```python
veces = 0 # esto me indica que las veces van a partir desde 0, python cuenta desde la posicion 0

while veces < 5:
    print("Hola")
    veces += 1
```



**Enunciado:** Suma los números del 1 al 100 y muestra el resultado.

**Solución:**

```python
numero = 1 # Definimos el numero
suma = 0 # Aqui guardaremos la suma acumulada

while numero <= 100: # mientras el numero sea menor o igual a 100
    suma = suma + numero # la suma sera el numero + suma
    numero += 1 # pasamos al siguiente numero de la iteracion hasta que se cumpla la condicion de que el numnero sea menor o igual a 100

print("La suma del 1 al 100 es:", suma)
```

> 💡 La variable `suma` es un **acumulador**: empieza en 0 y en cada vuelta se le agrega el número actual.


**Enunciado:** Pide una contraseña y sigue preguntando hasta que escriba `python123`. Cuando acierte, muestra "Acceso concedido".

```python
clave = ""

while clave != "python123":
    clave = input("Contraseña: ")
    if clave != "python123":
        print("Incorrecta, intenta de nuevo.")

print("Acceso concedido")
```

> 💡 `!=` significa "distinto de". La variable empieza vacía para poder entrar al ciclo la primera vez.


**Enunciado:** Pide números y súmalos hasta que el usuario escriba 0. Al final muestra la suma total.

```python
suma = 0
numero = int(input("Escribe un número (0 para terminar): "))

while numero != 0:
    suma = suma + numero
    numero = int(input("Escribe un número (0 para terminar): "))

print("La suma total es:", suma)
```


**Enunciado:** Pide números hasta que el usuario escriba 0 y al final muestra cuántos números ingresó (sin contar el 0).

```python
cantidad = 0
numero = int(input("Escribe un número (0 para terminar): "))

while numero != 0:
    cantidad += 1
    numero = int(input("Escribe un número (0 para terminar): "))

print("Ingresaste", cantidad, "números")
```

> 💡 Ahora el acumulador no suma los números: suma 1 por cada número ingresado. Esto se llama **contador**.


**Enunciado:** Pide una nota del 1.0 al 7.0. Si escribe algo fuera de ese rango, vuelve a preguntar hasta que sea válida.

```python
nota = float(input("Ingresa una nota (1.0 a 7.0): "))

while nota < 1.0 or nota > 7.0:
    print("Nota inválida.")
    nota = float(input("Ingresa una nota (1.0 a 7.0): "))

print("Nota registrada:", nota)
```

> 💡 `float(...)` se usa en vez de `int(...)` porque la nota puede tener decimales. `or` significa que basta con que una de las dos condiciones se cumpla.


**Enunciado:** El número secreto es 7. El programa dice "más alto" o "más bajo" hasta que el usuario acierte. Al final muestra cuántos intentos tomó.

```python
secreto = 7
intentos = 0

while True:
    intento = int(input("Adivina el número (1 al 10): "))
    intentos += 1

    if intento == secreto:
        print("¡Correcto! Lo lograste en", intentos, "intentos")
        break
    elif intento < secreto:
        print("Más alto")
    else:
        print("Más bajo")
```

> 💡 `while True` repite para siempre; `break` es lo que nos saca del ciclo cuando acertamos.
