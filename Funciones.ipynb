## 1. ¿Por qué necesitamos funciones?

Imagina que tienes que saludar a tres estudiantes con un mensaje decorado:

````python
print("********************")
print("Hola, Camila")
print("Bienvenida a INFO058")
print("********************")

print("********************")
print("Hola, Matías")
print("Bienvenida a INFO058")
print("********************")

print("********************")
print("Hola, Javiera")
print("Bienvenida a INFO058")
print("********************")
````

**Problemas de este código:**

- Repetimos lo mismo 3 veces. ¿Y si fueran 80 estudiantes?
- Si queremos cambiar los asteriscos por guiones, hay que cambiarlo en **todas** partes.
- Hay un error escondido: a Matías le dijimos "Bienvenid**a**". Copiar y pegar genera errores.

**Una función** es un bloque de código con nombre, que escribimos **una vez** y usamos **muchas veces**.

### Analogía: la receta
Una función es como una **receta de cocina**:

| Receta | Función |
|---|---|
| Nombre de la receta ("Sopaipillas") | Nombre de la función |
| Ingredientes | Parámetros (lo que recibe) |
| Pasos de preparación | Cuerpo de la función (el código) |
| El plato terminado | Valor de retorno (`return`) |

Escribir la receta **no** cocina nada. Recién cuando alguien la **sigue**, se obtiene el plato. Lo mismo pasa con las funciones: **definirla** no la ejecuta; hay que **llamarla**.

### ¡Ya han usado funciones!

````python
print("print es una función")
largo = len("Valdivia")     # len es una función: recibe un texto, devuelve un número
print(largo)
numero = int("42")          # int es una función: convierte texto a entero
print(numero + 1)
````

Hoy vamos a aprender a **crear nuestras propias funciones**.


## 2. Nuestra primera función

La estructura básica es:

```python
def nombre_de_la_funcion():
    # cuerpo: instrucciones (con sangría de 4 espacios)
    instruccion_1
    instruccion_2
```

Partes importantes:
- `def` → palabra clave que significa *define* (definir).
- El nombre → en minúsculas, con `_` entre palabras (igual que las variables).
- `()` → paréntesis, obligatorios.
- `:` → dos puntos al final, obligatorios.
- **Sangría (indentación)** → todo lo que está "corrido hacia adentro" pertenece a la función.

````python
def saludar():
    print("¡Hola!")
    print("Bienvenidos a la clase de funciones")
````

La función `saludar()` siempre hace lo mismo. Sería más útil si pudiera saludar a **cualquier persona**. Para eso usamos **parámetros**: variables que la función recibe entre los paréntesis.

````python
def saludar_a(nombre):
    print("Hola,", nombre)

saludar_a("Camila")
saludar_a("Matías")
saludar_a("Javiera")
````

### Vocabulario: parámetro vs argumento
- **Parámetro**: la variable en la **definición** → `nombre` en `def saludar_a(nombre):`
- **Argumento**: el valor concreto que se entrega en la **llamada** → `"Camila"` en `saludar_a("Camila")`

Cuando llamamos `saludar_a("Camila")`, Python hace internamente `nombre = "Camila"` y luego ejecuta el cuerpo.

## 4. `return`: que la función **devuelva** un resultado

Hasta ahora nuestras funciones solo **imprimen**. Pero muchas veces queremos que la función **calcule algo y nos entregue el resultado** para seguir usándolo. Para eso existe `return`.

````python
def cuadrado(x):
    return x * x

resultado = cuadrado(4)
print(resultado)
print(cuadrado(3) + cuadrado(4))   # podemos usar el resultado en otras operaciones
````

### 🔥 La diferencia más importante de la clase: `print` vs `return`

- `print` **muestra** algo en pantalla, para que lo vea un humano. El valor se "pierde".
- `return` **entrega** un valor al programa, para que se pueda guardar y seguir usando.

Analogía: `print` es que el cocinero **te muestre** el plato por la ventana. `return` es que **te lo entregue** para que te lo comas.

````python
def doble_print(x):
    print(x * 2)

z = doble_print(5)   # en pantalla aparece 10...
print(z)             # ...pero z vale None
````

````python
def doble(x):
    return x * 2

doble(5)                # el 10 se devuelve... y se pierde, nadie lo recibe
y = doble(5)            # el 10 se devuelve y TÚ lo guardas en y
print(doble(5) + 1)     # el 10 se usa directo en una operación, sin variable
````

🤔 **¿Por qué `a` vale `None`?**

Porque `doble_con_print` no tiene `return`. Toda función que no tiene `return` devuelve automáticamente `None` (que significa "nada"). El `10` que apareció en pantalla fue solo el `print` dentro de la función.

## 5. Parámetros con valor por defecto

Podemos darle a un parámetro un valor que se usa **si no se entrega argumento**.

````python
def saludar_con(nombre, saludo="Hola"):
    return saludo + ", " + nombre + "!"

print(saludar_con("Pedro"))               # usa el valor por defecto
print(saludar_con("Pedro", "Buenos días")) # reemplaza el valor por defecto
````

⚠️ Regla: los parámetros con valor por defecto van **al final**. `def f(a=1, b):` da error.

## 6. Alcance (*scope*): ¿dónde "vive" una variable?

Las variables creadas **dentro** de una función son **locales**: solo existen mientras la función se ejecuta.

````python
def calcular():
    resultado_interno = 100
    return resultado_interno

calcular()
# print(resultado_interno)   # descomenta: NameError, la variable no existe fuera
````

Lo que pasa en la función, se queda en la función. La forma correcta de "sacar" un valor es con `return`:

````python
valor = calcular()
print(valor)
````

### Mismo nombre, distinta variable

````python
x = 10

def cambiar():
    x = 99          # esta x es LOCAL, distinta de la x de afuera
    print("Dentro:", x)

cambiar()
print("Fuera:", x)
````

## 8. Funciones que usan otras funciones

La gran ventaja: podemos construir funciones grandes a partir de funciones pequeñas, como piezas de LEGO.

````python
def promedio(lista):
    """Devuelve el promedio de una lista de números."""
    return sum(lista) / len(lista)

def aprobado(nota):
    """Devuelve True si la nota es mayor o igual a 4.0."""
    return nota >= 4.0

def informe(nombre, notas):
    """Imprime el informe de un estudiante."""
    prom = promedio(notas)
    if aprobado(prom):
        estado = "APROBADO"
    else:
        estado = "REPROBADO"
    print(f"{nombre}: promedio {prom:.1f} → {estado}")

informe("Camila", [5.5, 6.0, 4.8])
informe("Matías", [3.2, 4.1, 3.0])
````

💡 Observa que `informe` no necesita saber **cómo** se calcula el promedio; solo confía en que `promedio` lo hace bien. Esto se llama **abstracción** y es la idea central de las funciones: dividir un problema grande en problemas pequeños.

### Funciones + ciclos

````python
curso = {
    "Camila": [5.5, 6.0, 4.8],
    "Matías": [3.2, 4.1, 3.0],
    "Javiera": [6.8, 7.0, 6.5],
}

for nombre in curso:
    informe(nombre, curso[nombre])
````

## 9. Resumen

```python
def nombre_funcion(param1, param2="valor_defecto"):
    """Qué hace la función."""
    # cuerpo (con sangría)
    resultado = param1 + param2
    return resultado          # devuelve el valor (opcional)

x = nombre_funcion(3, 4)      # llamada
```

| Concepto | Idea clave |
|---|---|
| `def` | Define la función (escribe la receta), no la ejecuta |
| Llamada `f()` | Ejecuta la función |
| Parámetro | Variable en la definición |
| Argumento | Valor concreto en la llamada |
| `return` | Entrega un valor y termina la función |
| Sin `return` | La función devuelve `None` |
| `print` ≠ `return` | Mostrar ≠ entregar |
| Variable local | Solo existe dentro de la función |

````python
# Ejercicio 1
def es_mayor_de_edad(edad):
    return edad >= 18

# Ejercicio 2
def contar_vocales(texto):
    contador = 0
    for letra in texto.lower():
        if letra in "aeiouáéíóú":
            contador += 1
    return contador

# Ejercicio 3
def mayor_de_tres(a, b, c):
    mayor = a
    if b > mayor:
        mayor = b
    if c > mayor:
        mayor = c
    return mayor

# Ejercicio 4
def precio_final(precio, descuento=0):
    con_descuento = precio * (1 - descuento / 100)
    return round(con_descuento * 1.19, 2)

# Ejercicio 5
def es_primo(n):
    if n < 2:
        return False
    for d in range(2, int(n ** 0.5) + 1):
        if n % d == 0:
            return False
    return True

def primos_hasta(n):
    primos = []
    for i in range(2, n + 1):
        if es_primo(i):
            primos.append(i)
    return primos

print(es_mayor_de_edad(17), es_mayor_de_edad(18))
print(contar_vocales("Universidad Austral"))
print(mayor_de_tres(4, 9, 2))
print(precio_final(10000, 10))
print(primos_hasta(30))
````

````python
# Desafío: validación de RUT
def calcular_dv(cuerpo):
    """Calcula el dígito verificador de un RUT (cuerpo como texto)."""
    suma = 0
    factor = 2
    for digito in reversed(cuerpo):
        suma += int(digito) * factor
        factor += 1
        if factor > 7:
            factor = 2
    resultado = 11 - (suma % 11)
    if resultado == 11:
        return "0"
    if resultado == 10:
        return "K"
    return str(resultado)

def validar_rut(rut):
    """Devuelve True si el RUT (formato 'cuerpo-dv') es válido."""
    rut = rut.replace(".", "").upper()
    cuerpo, dv = rut.split("-")
    return calcular_dv(cuerpo) == dv

print(validar_rut("12345678-5"))
print(validar_rut("12.345.678-0"))
````
