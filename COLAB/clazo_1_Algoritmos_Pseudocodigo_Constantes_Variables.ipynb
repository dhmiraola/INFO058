# Parte 1 — Algoritmos en Pseudocódigo

## 1.1 ¿Qué es un algoritmo?

Un **algoritmo** es una secuencia finita de instrucciones **exactas** y **ordenadas** que resuelve un problema.

Tres propiedades esenciales:

| Propiedad | Significado |
|---|---|
| **Precisión** | Cada instrucción es inequívoca (no admite interpretación) |
| **Orden** | La secuencia de pasos importa |
| **Finitud** | El algoritmo debe terminar en algún momento |

## 1.2 Instrucción exacta vs. instrucción ambigua

| Ambigua ❌ | Exacta ✅ |
|---|---|
| "Agrega azúcar al gusto" | "Agrega 2 cucharaditas de azúcar y revuelve 10 veces" |
| "Camina un poco hacia adelante" | "Camina 5 pasos hacia adelante" |

Un computador **solo** puede ejecutar instrucciones exactas. Si una instrucción admite más de una interpretación, el programa fallará o dará resultados inesperados.

## 1.3 Estructuras de control en pseudocódigo

Todo algoritmo se construye combinando solo tres estructuras:

- **Secuencia (S):** un paso después de otro
- **Condición (C):** "si... entonces... si no..."
- **Repetición (R):** "mientras... repetir..." / "repetir N veces..."


### 🧠 Ejemplo resuelto — Promedio de 3 notas

**Pseudocódigo:**
```
INICIO
  1. Leer nota1
  2. Leer nota2
  3. Leer nota3
  4. promedio = (nota1 + nota2 + nota3) / 3
  5. Mostrar promedio
FIN
```
Observa que cada paso es una acción concreta y el orden es indispensable: no se puede calcular el promedio (paso 4) antes de leer las notas (pasos 1–3).

# Parte 2 — Python: Variables, Constantes y Secuencias Simples

## 2.1 ¿Qué es una variable?

Una **variable** es un espacio en memoria con un **nombre** que almacena un valor. En Python **no se declara el tipo**: se infiere automáticamente del valor que le asignas.


````python
nombre = "Ana"      # str   (cadena de texto)
edad = 20           # int   (entero)
altura = 1.65        # float (número decimal)
es_estudiante = True  # bool  (verdadero / falso)

print(nombre, edad, altura, es_estudiante)
print(type(nombre), type(edad), type(altura), type(es_estudiante))
````

## 2.2 Reglas para nombrar variables

- Deben empezar con una letra o guion bajo `_` (nunca con un número)
- Pueden contener letras, números y guiones bajos
- Distinguen mayúsculas de minúsculas (`Edad` ≠ `edad`)
- No pueden ser palabras reservadas de Python (`if`, `for`, `class`, etc.)
- Por convención, se usa `snake_case`: `mi_variable`, `promedio_final`


````python
# ✅ Nombres válidos y recomendados
promedio_final = 6.2
_temporal = 10
edad2 = 19

# ❌ Esto generaría un error si lo descomentas:
# 2edad = 19        # no puede empezar con número
# mi-variable = 5   # el guion medio no está permitido

print(promedio_final, _temporal, edad2)
````

## 2.3 La función `print()`

`print()` muestra valores en pantalla. Puede recibir múltiples argumentos separados por coma, y se pueden combinar texto y variables usando **f-strings** (`f"..."`).


````python
nombre = "Ana"
edad = 20

print("Hola", nombre)                 # combinar strings y variables con coma
print("Edad:", edad, "años")
print(f"Hola {nombre}, tienes {edad} años")   # f-string (recomendado)
````

## 2.4 Constantes

Python **no tiene un tipo de dato "constante"** como otros lenguajes (no existe una palabra reservada `const`). Sin embargo, existe una **convención universal**: las variables que no deben cambiar de valor durante la ejecución se escriben en **MAYÚSCULAS**.

Esto es solo una convención de estilo — Python no impide que "la constante" cambie — pero cualquier programador que lea `PI = 3.1416` sabrá que **no debe modificarla**.


````python
# Constantes (por convención, en MAYUSCULAS)
PI = 3.1416
IVA = 0.19
GRAVEDAD = 9.8
NOMBRE_APP = "Gestor de Notas"

radio = 5
area_circulo = PI * radio ** 2
print("El área del círculo es: {area_circulo}", {area_circulo})
print("El valor de la gravedad es: {GRAVEDAD}", {area_circulo})
print()
````

## 2.5 Secuencias simples de instrucciones

Un programa en Python, en su forma más básica, es una **secuencia** de instrucciones que se ejecutan una tras otra, de arriba hacia abajo. Esto es exactamente la estructura de **secuencia (S)** que vimos en la Parte 1, ahora escrita en código real.

### 🧠 Ejemplo resuelto — de pseudocódigo a Python

**Pseudocódigo:**
```
INICIO
  1. Leer nota1
  2. Leer nota2
  3. Leer nota3
  4. promedio = (nota1 + nota2 + nota3) / 3
  5. Mostrar promedio
FIN
```

**Traducción a Python:**


````python
nota1 = 6.5
nota2 = 5.8
nota3 = 7.0

promedio = (nota1 + nota2 + nota3) / 3

print(f"El promedio es: {promedio}")
````

## 2.6 Entrada de datos con `input()`

`input()` permite que el usuario ingrese un valor durante la ejecución. **Siempre devuelve texto (`str`)**, por lo que si necesitas un número debes convertirlo con `int()` o `float()`.


````python
nombre = input("¿Cuál es tu nombre? ")
edad_texto = input("¿Cuántos años tienes? ")
edad = int(edad_texto)   # convertimos el texto a entero

print(f"Hola {nombre}, el próximo año tendrás {edad + 1} años")
````
