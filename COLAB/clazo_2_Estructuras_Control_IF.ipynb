````python
x = 5
if x > 0:
    if x % 2 == 0:
        print("Naranja")
    else:
        print("Pera")
else:
    print("Manzana")
````

````python
x = "hello"
y = "world"
if x == y:
    print(" x, y,  son iguales")
else:
    print("Son diferentes")
````

````python
x = 3
y = 7
z = 5
if x < y and y > z:
    if z > x:
        print("x < z < y")
    elif z < x:
        print("z < x < y")
    else:
        print("x == z < y")
else:
    print("No se cumple ninguna condición")
````

## 2. `if` / `else`

```
if condicion:
    # se ejecuta si condicion es True
else:
    # se ejecuta si condicion es False
```


````python
edad = int(input("que edad tienes"))

if edad >= 18:
    print("Eres mayor de edad")
else:
    print("Eres menor de edad")
````

## 3. `if` / `elif` / `else`

```
if condicion_1:
    # ...
elif condicion_2:
    # ...
else:
    # ...
```

**Regla fundamental:** Python evalúa las condiciones **en orden** y ejecuta solo el **primer** bloque cuya condición sea `True`. El resto se ignora, aunque también sean verdaderas.


````python
nota = 75

if nota >= 90:
    print("A")
elif nota >= 80:
    print("B")
elif nota >= 70:
    print("C")
elif nota >= 60:
    print("D")
else:
    print("F")
````

## 4. Operadores lógicos: `and`, `or`, `not`

| Operador | Verdadero cuando... |
|---|---|
| `and` | **ambas** condiciones son `True` |
| `or` | **al menos una** condición es `True` |
| `not` | invierte el valor lógico |


````python
edad = 20
tiene_carnet = True

puede_arrendar_auto = edad >= 18 and tiene_carnet
print(puede_arrendar_auto)

nota = -5
nota_invalida = nota < 0 or nota > 100
print(nota_invalida)

esta_aprobado = False
print(not esta_aprobado)
````

## 5. Condicionales anidados

Se puede colocar un `if` dentro de otro `if` cuando hay más de un nivel de decisión.


````python
usuario_registrado = True
password_correcta = False

if usuario_registrado:
    if password_correcta:
        print("Acceso concedido")
    else:
        print("Contraseña incorrecta")
else:
    print("Usuario no registrado")
````
