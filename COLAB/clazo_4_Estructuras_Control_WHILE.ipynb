## 1. Estructura de `while`

```python
while condicion:
    # se repite mientras condicion sea True
```

⚠️ **Peligro:** si la condición nunca se vuelve `False` y no hay `break`, el bucle corre **infinitamente**. Siempre asegúrate de que algo dentro del bucle modifique la condición.


````python
contador = 0

while contador < 5: # Mientras el contador es menor que cinco
    print(contador)
    contador = contador + 1   # sin esto, el bucle seria infinito

print("Fin del bucle")
````

## 2. `break`: salir inmediatamente del bucle


````python
numero = 1

while True:            # bucle "infinito" controlado con break
    print(numero)
    if numero == 5:
        break           # sale del bucle al llegar a 5
    numero = numero + 1
````

## 3. `continue`: saltar a la siguiente iteración


````python
numero = 0

while numero < 10:
    numero = numero + 1

    if numero % 2 == 0:  # Si es par...
        continue        # vuelve al while sin llegar al print

    print(numero)       # Solo llegan aquí los impares
````

## 4. Equivalencia `for` ↔ `while`

Todo `for` se puede reescribir como `while`, controlando manualmente el contador.

**Con `for`:**
```python
for i in range(1, 6):
    print(i)
```

**Equivalente con `while`:**


````python
i = 1
while i < 6:
    print(i)
    i = i + 1
````
