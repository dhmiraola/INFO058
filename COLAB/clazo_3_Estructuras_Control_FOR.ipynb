## 1. `for` con `range()`

```python
range(inicio, fin, paso)
```
El valor `fin` **nunca se incluye**.


````python
for i in range(5,10):        # i toma valores: 0, 1, 2, 3, 4
    print(i)
````

````python
for i in range(1, 6):     # i toma valores: 1, 2, 3, 4, 5
    print(i)
````

````python
for i in range(0, 10, 2):  # i toma valores: 0, 2, 4, 6, 8
    print(i)
````

### 🔍 Tracing — completa antes de ejecutar

```python
total = 0
for i in range(1, 5):
    total = total + i
    print(i, total)
```

| Iteración | i | total |
|---|---|---|
| 1 | | |
| 2 | | |
| 3 | | |
| 4 | | |

Escribe tu predicción arriba, luego ejecuta la celda siguiente y compara.


````python
total = 0
for i in range(1, 5):
    total = total + i
    print(i, total)
````

## 2. Recorrer una lista o un string con `for`

`for` también sirve para recorrer directamente los elementos de una secuencia, sin necesidad de índices.


````python
frutas = ["manzana", "pera", "uva"]

for fruta in frutas:
    print(fruta)

print("---")

for letra in "Python":
    print(letra)
````

## 3. Acumuladores y contadores

Un patrón muy común: usar una variable externa al bucle para ir acumulando un resultado (suma, conteo, producto, etc).


````python
# Suma de los primeros 10 numeros
suma = 0
for i in range(1, 11):
    suma = suma + i
print("La suma es:", suma)

# Contar cuantos numeros pares hay del 1 al 20
contador = 0
for i in range(1, 21):
    if i % 2 == 0:
        contador = contador + 1
print("Cantidad de pares:", contador)
````
